```python
print("Orders duplicates:", olist_orders_dataset.duplicated().sum())
print("Items duplicates:", olist_orders_items.duplicated().sum())
print("Reviews duplicates:", olist_orders_reviews.duplicated().sum())


olist_orders_dataset = olist_orders_dataset.drop_duplicates()
olist_orders_items = olist_orders_items.drop_duplicates()
olist_orders_reviews = olist_orders_reviews.drop_duplicates()


print("Orders duplicates after:", olist_orders_dataset.duplicated().sum())
print("Items duplicates after:", olist_orders_items.duplicated().sum())
print("Reviews duplicates after:", olist_orders_reviews.duplicated().sum())
```

```python
data = pd.merge(
    olist_orders_dataset,
    olist_orders_items,
    on = "order_id",
    how = "left"
)
```

```python
data
```

```python
final_data = pd.merge(
    olist_orders_reviews,
    data,
    on = "order_id",
    how = "left"
    
)
```
